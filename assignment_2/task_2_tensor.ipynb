### 2.1. Создание тензора

```python
# Отдельный seed
import numpy as np
np.set_printoptions(precision=2, suppress=True, threshold=1000)
rng_temperatures = np.random.default_rng(123)
T = rng_temperatures.uniform(-10, 35, (7, 24, 3))
print("shape:", T.shape, "ndim:", T.ndim, "size:", T.size, "dtype:", T.dtype)
print("Дни:", T.shape[0], "часы:", T.shape[1], "города:", T.shape[2])
assert T.shape == (7, 24, 3) and ((T >= -10) & (T <= 35)).all()
```

### 2.2. Индексы и срезы

```python
# Индексы начинаются с 0
second_city = T[:, :, 1]
first_day = T[0, :, :]
daytime = T[:, 9:19, :]
weekend_first_city = T[5:7, :, 0]
eight_am = T[:, 8, :]
for label, values in [("Второй город, все дни и часы", second_city),
                      ("Первый день, все города", first_day),
                      ("09:00–18:00 включительно, все дни и города", daytime),
                      ("Первый город, дни 6 и 7", weekend_first_city),
                      ("08:00, все дни и города", eight_am)]:
    print(label, "shape:", values.shape, "температуры (°C):")
    print(values)
print("День 3, 12:00, город 2 (°C):", T[2, 12, 1])
assert daytime.shape == (7, 10, 3) and eight_am.shape == (7, 3)
```

### 2.3. Агрегация по осям

```python
# axis указывает, какие оси убрать при подсчёте среднего или максимума
city_mean = T.mean(axis=(0, 1))
city_min = T.min(axis=(0, 1))
city_max = T.max(axis=(0, 1))
daily_max = T.max(axis=(1, 2))
daily_city_mean = T.mean(axis=1)
hour_mean = T.mean(axis=(0, 2))
day_mean = T.mean(axis=(1, 2))
for label, values in [("Среднее по городам, axis=(0,1)", city_mean),
                      ("Минимум по городам, axis=(0,1)", city_min),
                      ("Максимум по городам, axis=(0,1)", city_max),
                      ("Максимум по дням, axis=(1,2)", daily_max),
                      ("Среднее по дням и городам, axis=1", daily_city_mean),
                      ("Среднее по часам, axis=(0,2)", hour_mean),
                      ("Среднее по дням, axis=(1,2)", day_mean)]:
    print(label, "shape:", values.shape, "значения (°C):", values)
print("Час с минимальным средним:", hour_mean.argmin(), "среднее (°C):", hour_mean.min())
print("Самый тёплый день (с 1):", day_mean.argmax() + 1, "среднее (°C):", day_mean.max())
print("Самый тёплый город (с 1):", city_mean.argmax() + 1, "среднее (°C):", city_mean.max())
assert daily_city_mean.shape == (7, 3)
```

### 2.4. Изменение формы

```python
# Меняем только форму
temperature_matrix = T.reshape(7, 72)
restored_tensor = temperature_matrix.reshape(7, 24, 3)
print("Форма матрицы:", temperature_matrix.shape)
print("Восстановленная форма:", restored_tensor.shape)
print("Полное совпадение:", np.array_equal(T, restored_tensor))
assert np.array_equal(T, restored_tensor)
```

### 2.5. Корреляция городов

```python
# Сначала выносим города вперёд
city_sequences = T.transpose(2, 0, 1).reshape(3, 168)
city_correlations = np.corrcoef(city_sequences)
print("Форма последовательностей городов:", city_sequences.shape)
print("Матрица корреляций:", np.round(city_correlations, 4))
city_pair_rows, city_pair_cols = np.triu_indices(3, k=1)
best_pair_index = np.argmax(city_correlations[city_pair_rows, city_pair_cols])
best_city_1 = city_pair_rows[best_pair_index]
best_city_2 = city_pair_cols[best_pair_index]
print(f"Максимальная корреляция разных городов: {best_city_1 + 1} и {best_city_2 + 1}, "
      f"r = {city_correlations[best_city_1, best_city_2]:.4f}")
assert np.array_equal(city_sequences[1], T[:, :, 1].ravel())
```
