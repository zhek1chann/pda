# Task 1 — NumPy
Случайные данные; цены — млн тенге.

### 1.1. Массивы

```python
# Seed повторяет результат
import numpy as np
np.set_printoptions(precision=2, suppress=True, threshold=1000)
rng_apartments = np.random.default_rng(42)
area_array = rng_apartments.uniform(25, 150, 200)
rooms_array = rng_apartments.integers(1, 6, 200)
price_array = np.round(rng_apartments.uniform(15, 120, 200), 2)
for name, values in [("Площадь (м²)", area_array), ("Комнаты", rooms_array),
                     ("Цена (млн тенге)", price_array)]:
    print(name, "shape:", values.shape, "ndim:", values.ndim,
          "size:", values.size, "dtype:", values.dtype)
assert area_array.shape == rooms_array.shape == price_array.shape == (200,)
assert ((area_array >= 25) & (area_array <= 150)).all()
assert ((rooms_array >= 1) & (rooms_array <= 5)).all()
assert ((price_array >= 15) & (price_array <= 120)).all()
```

`shape` — форма, `ndim` — оси, `size` — размер, `dtype` — тип. Один индекс — одна квартира.

### 1.2. Срезы и маски

```python
# Срезы показывают часть данных
for name, values in [("Площадь (м²)", area_array), ("Комнаты", rooms_array),
                     ("Цена (млн тенге)", price_array)]:
    print(name, "первые 10:", values[:10], "последние 10:", values[-10:])
print("Площади с индексами 20–39 (м²):", area_array[20:40])
large_mask = area_array > 70
above_mean_mask = price_array > price_array.mean()
large_family_mask = (area_array > 70) & (rooms_array >= 3)
budget_family_mask = ((rooms_array >= 2) & (rooms_array <= 4)
                     & (area_array >= 60) & (area_array <= 100) & (price_array <= 80))
print("Площадь > 70 м², количество:", large_mask.sum())
print("Цена выше средней, количество:", above_mean_mask.sum(),
      "первые 10 цен (млн тенге):", price_array[above_mean_mask][:10])
print("Площадь > 70 м² и комнат >= 3, количество:", large_family_mask.sum())
print("2–4 комнаты, 60–100 м², до 80 млн тенге, количество:", budget_family_mask.sum())
print("Первые 10 подходящих индексов:", np.flatnonzero(budget_family_mask)[:10])
```

`[20:40]` — индексы 20–39. Маска выбирает `True`, `&` объединяет условия.

### 1.3. Статистика

```python
# Считаем центр, разброс и процентили, затем находим квартиры-экстремумы
percentile_levels = [10, 25, 50, 75, 90, 99]
for name, values in [("Площадь (м²)", area_array), ("Цена (млн тенге)", price_array)]:
    print(name)
    print("Среднее, медиана, минимум, максимум, ст. отклонение, дисперсия:",
          np.array([values.mean(), np.median(values), values.min(), values.max(),
                    values.std(), values.var()]))
    print("Процентили", percentile_levels, ":", np.percentile(values, percentile_levels))
for label, index in [("Самая дорогая", np.argmax(price_array)),
                     ("Самая большая", np.argmax(area_array))]:
    print(label, "индекс:", index, "цена (млн тенге):", price_array[index],
          "площадь (м²):", area_array[index], "комнаты:", rooms_array[index])
print("Цен ниже медианы:", (price_array < np.median(price_array)).sum())
print("Площадей выше среднего:", (area_array > area_array.mean()).sum())
print(f"Средняя цена {price_array.mean():.2f}, медиана {np.median(price_array):.2f}, "
      f"их разница {price_array.mean() - np.median(price_array):.2f} млн тенге.")
```

Разница среднего и медианы может указывать на выбросы. Здесь данные случайные; дисперсия — квадрат стандартного отклонения.

### 1.4. Новые признаки

```python
# Делим массивы поэлементно
price_per_m2_array = price_array / area_array
price_per_m2_kzt_array = price_per_m2_array * 1_000_000
print("Форма массива цен за м²:", price_per_m2_array.shape)
print("Среднее, медиана, минимум, максимум (млн тенге/м²):",
      [price_per_m2_array.mean(), np.median(price_per_m2_array),
       price_per_m2_array.min(), price_per_m2_array.max()])
print("Первые пять (тенге/м²):", price_per_m2_kzt_array[:5])
highest_unit_price_index = np.argmax(price_per_m2_array)
print("Максимальная цена за м²: индекс", highest_unit_price_index,
      "цена (млн тенге)", price_array[highest_unit_price_index],
      "площадь (м²)", area_array[highest_unit_price_index],
      "rooms", rooms_array[highest_unit_price_index],
      "цена за м² (млн тенге/м²)", price_per_m2_array[highest_unit_price_index])
print("Цена за м² выше средней, количество:", (price_per_m2_array > price_per_m2_array.mean()).sum())
expensive_flag = price_array > np.percentile(price_array, 75)
print("75-й процентиль (млн тенге):", np.percentile(price_array, 75))
print("Тип флага дорогой квартиры:", expensive_flag.dtype,
      "True:", expensive_flag.sum(), "False:", (~expensive_flag).sum())
assert np.isfinite(price_per_m2_array).all()
assert expensive_flag.shape == (200,) and expensive_flag.dtype == bool
```

Цена за м² — деление без цикла. `expensive_flag` — цена выше 75-го процентиля.

### 1.5. Нормализация

```python
# Вычитаем минимум и делим на размах, чтобы получить масштаб [0, 1]
area_normalized = (area_array - area_array.min()) / (area_array.max() - area_array.min())
price_normalized = (price_array - price_array.min()) / (price_array.max() - price_array.min())
for name, original, normalized in [("Площадь", area_array, area_normalized),
                                   ("Цена", price_array, price_normalized)]:
    print(name, "границы после нормализации:", normalized.min(), normalized.max())
    print("Первые пять: исходное | нормализованное")
    print(np.column_stack((original[:5], normalized[:5])))
    assert np.isclose(normalized.min(), 0) and np.isclose(normalized.max(), 1)
```

Min-Max даёт [0, 1]. Удобно сравнивать разные признаки. При `max == min` делить нельзя.
