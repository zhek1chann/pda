### 1.1. Массивы

```python
import numpy as np
np.set_printoptions(precision=2, suppress=True, threshold=1000)
rng_apartments = np.random.default_rng(42)
area_array = rng_apartments.uniform(25, 150, 200)
rooms_array = rng_apartments.integers(1, 6, 200)
price_array = np.round(rng_apartments.uniform(15, 120, 200), 2)
for name, values in [("Area (m²)", area_array), ("Rooms", rooms_array),
                     ("Price (million KZT)", price_array)]:
    print(name, "shape:", values.shape, "ndim:", values.ndim,
          "size:", values.size, "dtype:", values.dtype)
assert area_array.shape == rooms_array.shape == price_array.shape == (200,)
assert ((area_array >= 25) & (area_array <= 150)).all()
assert ((rooms_array >= 1) & (rooms_array <= 5)).all()
assert ((price_array >= 15) & (price_array <= 120)).all()
```

### 1.2. Срезы и маски

```python
for name, values in [("Area (m²)", area_array), ("Rooms", rooms_array),
                     ("Price (million KZT)", price_array)]:
    print(name, "first 10:", values[:10], "last 10:", values[-10:])
print("Areas at indices 20–39 (m²):", area_array[20:40])
large_mask = area_array > 70
above_mean_mask = price_array > price_array.mean()
large_family_mask = (area_array > 70) & (rooms_array >= 3)
budget_family_mask = ((rooms_array >= 2) & (rooms_array <= 4)
                     & (area_array >= 60) & (area_array <= 100) & (price_array <= 80))
print("Area > 70 m² count:", large_mask.sum())
print("Price above mean count:", above_mean_mask.sum(),
      "first 10 prices (million KZT):", price_array[above_mean_mask][:10])
print("Area > 70 m² and rooms >= 3 count:", large_family_mask.sum())
print("2–4 rooms, 60–100 m², up to 80 million KZT count:", budget_family_mask.sum())
print("First 10 matching indices:", np.flatnonzero(budget_family_mask)[:10])
```

### 1.3. Статистика

```python
percentile_levels = [10, 25, 50, 75, 90, 99]
for name, values in [("Area (m²)", area_array), ("Price (million KZT)", price_array)]:
    print(name)
    print("Mean, median, min, max, standard deviation, variance:",
          np.array([values.mean(), np.median(values), values.min(), values.max(),
                    values.std(), values.var()]))
    print("Percentiles", percentile_levels, ":", np.percentile(values, percentile_levels))
for label, index in [("Most expensive", np.argmax(price_array)),
                     ("Largest area", np.argmax(area_array))]:
    print(label, "index:", index, "price (million KZT):", price_array[index],
          "area (m²):", area_array[index], "rooms:", rooms_array[index])
print("Prices below median:", (price_array < np.median(price_array)).sum())
print("Areas above mean:", (area_array > area_array.mean()).sum())
print(f"Mean price {price_array.mean():.2f}, median {np.median(price_array):.2f}, "
      f"difference {price_array.mean() - np.median(price_array):.2f} million KZT.")
```

### 1.4. Новые признаки

```python
price_per_m2_array = price_array / area_array
price_per_m2_kzt_array = price_per_m2_array * 1_000_000
print("Unit price array shape:", price_per_m2_array.shape)
print("Mean, median, min, max (million KZT/m²):",
      [price_per_m2_array.mean(), np.median(price_per_m2_array),
       price_per_m2_array.min(), price_per_m2_array.max()])
print("First five (KZT/m²):", price_per_m2_kzt_array[:5])
highest_unit_price_index = np.argmax(price_per_m2_array)
print("Maximum unit price: index", highest_unit_price_index,
      "price (million KZT)", price_array[highest_unit_price_index],
      "area (m²)", area_array[highest_unit_price_index],
      "rooms", rooms_array[highest_unit_price_index],
      "unit price (million KZT/m²)", price_per_m2_array[highest_unit_price_index])
print("Unit price above mean count:", (price_per_m2_array > price_per_m2_array.mean()).sum())
expensive_flag = price_array > np.percentile(price_array, 75)
print("75th percentile (million KZT):", np.percentile(price_array, 75))
print("Expensive flag dtype:", expensive_flag.dtype,
      "True:", expensive_flag.sum(), "False:", (~expensive_flag).sum())
assert np.isfinite(price_per_m2_array).all()
assert expensive_flag.shape == (200,) and expensive_flag.dtype == bool
```

### 1.5. Нормализация

```python
area_normalized = (area_array - area_array.min()) / (area_array.max() - area_array.min())
price_normalized = (price_array - price_array.min()) / (price_array.max() - price_array.min())
for name, original, normalized in [("Area", area_array, area_normalized),
                                   ("Price", price_array, price_normalized)]:
    print(name, "normalized bounds:", normalized.min(), normalized.max())
    print("First five: original | normalized")
    print(np.column_stack((original[:5], normalized[:5])))
    assert np.isclose(normalized.min(), 0) and np.isclose(normalized.max(), 1)
```
